```python


def uniplot(df, col, title, hue=None, use_log=False):
    # 1. Styling Configurations
    sns.set_style('whitegrid')
    sns.set_context('talk')
    plt.rcParams["axes.labelsize"] = 14
    plt.rcParams["axes.titlesize"] = 18
    plt.rcParams["axes.titlepad"] = 20

    # 2. Correct Dynamic Width Calculation
    n_col_cats = df[col].nunique()
    n_hue_cats = df[hue].nunique() if (hue and hue in df.columns) else 1
    
    # Calculate appropriate width to avoid squishing/stretching
    width = max(8, (n_col_cats * n_hue_cats * 0.8) + 4)
    
    fig, ax = plt.subplots(figsize=(width, 6))

    # 3. Plot Countplot
    ax = sns.countplot(
        data=df, 
        x=col, 
        order=df[col].value_counts().index, 
        hue=hue, 
        palette='bright'
    )

    # 4. Conditional Log Scale
    if use_log:
        plt.yscale('log')
        plt.ylabel("Count (Log Scale)")
    else:
        plt.ylabel("Count")

    # 5. Add Exact Count Labels on Top of Each Bar
    for p in ax.patches:
        height = p.get_height()
        if pd.notnull(height) and height > 0:
            ax.annotate(
                f'{int(height)}',
                (p.get_x() + p.get_width() / 2., height),
                ha='center', va='bottom',
                fontsize=10, color='black',
                xytext=(0, 3), textcoords='offset points'
            )

    # 6. Formatting & Rotation
    plt.xticks(rotation=45, ha='right')
    plt.title(title, fontweight='bold')
    plt.tight_layout()
    
    plt.show()
```
